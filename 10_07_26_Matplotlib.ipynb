# ***Matplotlib:Lineplots:***
1. matplotlib.pyplot : Used to draw graphs in python.
2. plt.plot(x, y): by using x-axis & y-axis data to form a graph this functin we use.
3. plt.title(), plt.xlabel(), plt.ylabel() : is used to name graph, x,y axis names.
4. plt.show() : it used to show the graph at screen on final.



```python
import matplotlib.pyplot as plt
# Lab data: X-Axis: time, Y-Axis: temperature
time_min = [0, 5, 10, 15, 20, 25]
temperature_C = [20, 40,55, 75, 90,100]
# To draw lineplot
plt.plot(time_min, temperature_C, color='green', marker='o', linestyle='--')
# Adding professional information to graph(names to chart,axis-x,y)
plt.title("Chemical Reaction: Temperature vs Time")
plt.xlabel ("Time (minutes)")
plt.ylabel ("Temperature (°C)")
# Showing graph at final
print("--- Matplotlib Lab: Creating our First Line Plot ---")
plt.show()
```

Labels & Legends:

label="Chemical_name": by using this we can predict eachline in grapg by its chemical name.
plt.legend(): by using this function inside of our graoh small box open and indiactes each color line by its chemical name.

```python
import matplotlib.pyplot as plt
# X-axis: reaction time (minutes)
time_min = [0,5,10,15,20,25]
# Y-axis: taking two different chemicals temperatures
ZrO2_temp = [25, 40, 55, 75, 90, 100]
TiO2_temp = [25,35,45,60,75,85]
# Here we draw two different lines by diffrent color & label variation
plt.plot(time_min, ZrO2_temp, color='green', marker= 'o', label= 'ZrO2 Nanoparticles')
plt.plot(time_min, TiO2_temp, color='blue', marker= 's', label= 'TiO2 Nanoparticles')
# Graph professional data
plt.title("Comparitive Analysis: ZrO2 vs TiO2 Thermal Reaction")
plt.xlabel("Time (Minutes)")
plt.ylabel("Temperature (°C)")
# Showing labels box on screen (legend)
plt.legend()
# Showing final graph on the screen
print("---Matplotlib Lab: Multi_Line graph With Legends")
plt.show()
```

Custamizing plots:

plt.figure(figsize=(10,5)): It increases graph width=10,height=5 inches to make image big size to fit to PC.
fontsize=12: x,y axis titls sizes increase to read easyly.


```python
import matplotlib.pyplot as plt

time_min = [0, 5, 10, 15, 20, 25]
ZrO2_temp = [25, 40, 55, 75, 90, 100]
TiO2_temp = [25, 35, 45, 60, 75, 85]

# 1. గ్రాఫ్ సైజ్ ని పెద్దదిగా మార్చడం (ఇది అన్నింటికంటే పైన రాయాలి)
plt.figure(figsize=(10, 5))

# 2. రెండు లైన్స్ గీయడం
plt.plot(time_min, ZrO2_temp, color='black', marker='o', linewidth=2, label='ZrO2 Nanoparticles')
plt.plot(time_min, TiO2_temp, color='blue', marker='s', linewidth=2, label='TiO2 Nanoparticles')

# 3. టైటిల్ మరియు లేబుల్స్ యొక్క ఫాంట్ సైజ్ పెంచడం (fontsize)
plt.title("Comparative Analysis: ZrO2 vs TiO2 Thermal Reaction", fontsize=14, fontweight='bold')
plt.xlabel("Time (Minutes)", fontsize=12)
plt.ylabel("Temperature (°C)", fontsize=12)

# 4. లెజెండ్ మరియు గ్రిడ్ యాడ్ చేయడం
plt.legend(fontsize=10)
plt.grid(True, linestyle='--', alpha=0.5)

print("--- Matplotlib Lab: Customizing Size and Fonts ---")
plt.show()
```

*Seaborn*

import seaborn as sns
sns.scatterplot() : It shos the realtionship between two different features by dots
hue: this parameter mainly indicated colors accordind to each chemical in graph.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
data = {
    "Nanoparticle": ["ZrO2", "TiO2", "SiO2", "Al2O3", "ZnO"],
    "Size_nm": [25, 45, 60, 35, 50],
    "Reactivity": [85, 65, 40, 75, 55]
}
df = pd.DataFrame(data)
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='Size_nm', y= 'Reactivity',hue= 'Nanoparticle', s=150 )
plt.title("Nanoparticle Size vs Reactivity", fontsize=14, fontweight='bold')
plt.xlabel("Size (nm)", fontsize=12)
plt.ylabel("Reactivity Score", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)

print("--- Seaborn Lab: Creating a Scatter Plot ---")
plt.show()
```

seaborn barplot:

It is use to comparitive analysis of different properties by using barcharts.
sns.barplot(): This function gives comparitive analysis between catogorical data and numerical data.
palette:This parameter used to apply different colors to each bar in chart.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
chemical_data = {
    "Nanoparticles": ["ZrO2", "TiO2", "SiO2", "Al2O3"],
    "Size_nm": [25.0, 43.33, 45.0, 60.0]
}
df = pd.DataFrame(chemical_data)
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x='Nanoparticles', y='Size_nm', palette='viridis')
plt.title("Comparative Size Analysis of Nanoparticles", fontsize=14, fontweight='bold')
plt.xlabel("Nanoparticles", fontsize=12)
plt.ylabel("Size (nm)", fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.savefig("nanoparticle_barplot.png", dpi=300)

print("--- Seaborn Lab: Creating a Bar Plot ---")
plt.show()
```

Seaborn: Histograms

seaborn histograms used to ind range of distribution of data (out of thousands of nanaoparticles we can find range and distribution)
sns.histoplot(): Data distribution range can be visuvalize by using this function.
kde=True (kernel density estimate): Along with bars it shows curve which shows data trend.
bins(): this parameter is used to decide number of intervals of data.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
np.random.seed(42)
nanoparticle_sizes= np.random.normal(loc=45, scale=5, size=100)
df = pd.DataFrame({"Particle_Size_nm": nanoparticle_sizes})
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='Particle_Size_nm', kde=True, color='darkgreen', bins=15)
plt.title("Distribution Analysis of Synthesized Nanoparticle Sizes", fontsize=14, fontweight='bold')
plt.xlabel("Particle Size (nm)", fontsize=12)
plt.ylabel("Count / Frequency", fontsize=12)
plt.grid(axis='both', linestyle='--', alpha=0.5)
plt.savefig("nanoparticle_histogram.png", dpi=300)

print("--- Seaborn Lab: Creating a Histogram with KDE ---")
plt.show()
```

Heatmaps & Correlation:

It is used to show correlation of various parameters data in colorgrid form.
df.corr() : It shows mathematical correlation between varous columns of data. its values ranges from -1 to +1.
sns.heatmap() : It is used to show the correlation in colorful grid form.
annot=True(Annotations) : It is used to print that correlation numbers in the inside of that colorgrid boxes.
camp: It is a parameter is used to set the color combinations of the dataset (Ex: coolwarm, viridis, Blues)

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
lab_data = {
    " Microwave_Temp_C": [100, 120, 140, 160, 180, 200],
    "Reaction_Time_mins": [10, 15, 20, 25, 30, 35],
    "Particle_Size_nm": [25, 32, 40, 45, 53, 60],
    "Stability_Score": [95, 88, 82, 75, 68, 60]
}
df = pd.DataFrame(lab_data)

correlation_matrix = df.corr()

plt.figure(figsize=(8, 6))

sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', linewidths=0.5, fmt=".2f")

plt.title("Correlation Matrix of Green Synthesis Parameters", fontsize=14, fontweight='bold', pad=15)

plt.savefig("nanoparticle_heatmap.png", dpi=300, bbox_inches='tight')

print("--- Seaborn Lab: Generating a Professional Heatmap ---")
plt.show()
```
