Handling Missing Data in Chemistry Datasets:

Missing Values (NaN): In lab during recording reading some values may missed by sensor errors.
df.isnull().sum() : It find out in each cell how many null values are present.
df.fillna() : It fills null value by mean of its row values.

```python
import pandas as pd
import numpy as np
raw_lab_data = {
    "Microwave_Temp_C": [120, 140, np.nan, 160, 180, 200],
    "Reaction_Time_mins": [15, 20, 25, np.nan, 35, 40],
    "Particle_Size_nm": [32, 40, 45, 53, np.nan, 65]
}
df = pd.DataFrame(raw_lab_data)
print("--- 1. Original Dataset with Missing Values (NaN) ---")
print(df)
print("\n--- 2. Checking Count of Missing Values ---")
print(df.isnull().sum())

df_cleaned = df.copy()
for column in df_cleaned.columns:
    mean_value = df_cleaned[column].mean()
    df_cleaned[column] = df_cleaned[column].fillna(mean_value)

print("\n--- 3. Cleaned Dataset (NaN replaced with Column Mean) ---")
print(df_cleaned.round(2))
```

Outlier detection:

In lab due to voltage up and downs sometimes it may give abnormal values.
taht abnormal value called outlier, that is detected by Z-score value.

Outlier: Abnormal value generated due to human/mition errors.
Z-score: when lab values are more than +3/-3, then it is considered as outlier it can be removed from data to clean data and build the good ML model.

```python
import pandas as pd
import numpy as np
from scipy import stats

lab_data_with_outlier = {
    "Microwave_Temp_C": [120, 140, 150, 160, 170, 180, 190, 200],
    "Particle_Size_nm": [32, 35, 38, 42, 250, 48, 52, 55]
}

df_outlier = pd.DataFrame(lab_data_with_outlier)
print("--- 1. Dataset with Outlier (Look at index 4) ---")
print(df_outlier)
df_outlier['Z_Score'] = np.abs(stats.zscore(df_outlier['Particle_Size_nm']))

print("\n--- 2. Calculated Z-Scores ---")
print(df_outlier.round(2))
threshold = 2
df_filtered = df_outlier[df_outlier['Z_Score'] < threshold].drop(columns=['Z_Score'])

print(f"\n--- 3. Cleaned Dataset (Outlier Removed using Z-Score > {threshold}) ---")
print(df_filtered)
```

Feature Scaling:

 Min-Max Normalization

 Feature Scaling: Making different scsling data of coloumn in to one range scale

 .Min-Max Scaler: It changes datavalues from exactly zero to 1 range.
  
  Its formula:

 $$X_{scaled} = \frac{X - X_{min}}{X_{max} - X_{min}}$$

 ML Readiness: After completion of this process only our ML-Model is 100% ready to go for Linear Regression, Neural Networks.

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler


cleaned_lab_data = {
    "Microwave_Temp_C": [120.0, 140.0, 160.0, 160.0, 180.0, 200.0],
    "Reaction_Time_mins": [15.0, 20.0, 25.0, 27.0, 35.0, 40.0],
    "Particle_Size_nm": [32.0, 40.0, 45.0, 53.0, 47.0, 65.0]
}
df_ready = pd.DataFrame(cleaned_lab_data)

print("--- 1. Data Before Scaling (Original Units) ---")
print(df_ready)

scaler = MinMaxScaler()
scaled_features = scaler.fit_transform(df_ready)

df_scaled = pd.DataFrame(scaled_features, columns=df_ready.columns)

print("\n--- 2. Data After Min-Max Scaling (All Values Between 0 and 1) ---")
print(df_scaled.round(3))
```
