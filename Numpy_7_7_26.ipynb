2D array slicing process:
2D slicing means cutting specif row or coumn data completely out from array.
if we give [:,0] here colon means select all rows and 0 means only first coumn. means we are taking out data of all rows 1st coumn information.

```python
import numpy as np
chemical_table = [
    [25,45,60],
    [30,50,75]
]
two_d_array = np.array(chemical_table)
first_column_slice = two_d_array[:,0]
print("--Numpy lab:2-D slicing--")
print("our 2d total table:\n", two_d_array)
print("data of sliced first column:", first_column_slice)
```

Numpy mathematical operations:

In numpy when we have thousands of data and we need o calculate any mathematicla operations like addition, substraction,multiplication we can do it for entire data at a time, its called vectorisation. we can calculate the average of all data by using np.mean() functin in numpy.

```python
import numpy as np
# In one chemical batch 4 nanoparticles sizes--
nano_sizes = [20,35,50,65]
chemistry_array = np.array(nano_sizes)
# Addition: here we are going to add 5nm to each size
increased_sizes = chemistry_array + 5
# Average\Mean calculation:
average_size = np.mean(chemistry_array)
print("--Numpy lab: Mathamatical operations---")
print("Our particles original sizes:", chemistry_array)
print("5nm added sizes:",increased_sizes)
print("Average of this batch:", average_size)
```

PANDAS:
Pandas is a python libraty mainly used analyse big data and manipulation of data. pandas has 2 datastructures/: 1. series, 2. Dtaframe
1.Series: It contains only one column, (as like numpy 1D)  
2. Dtaframes: it contains rows & columns tablar formate data (as like np.2D table)

```python

import pandas as pd
# Three types of nanoparticles with sizes in dictionary data form--
chemical_data={
    "Nanoparticle": ["ZrO2", "TiO2", "SiO2"],
    "Size_nm" : [25, 45, 60],
    "stability": ["High", "Medium", "High"]
}
# Dta converted into pandas dataframe--
df = pd.DataFrame(chemical_data)
print("--Pandas lab: creating our first DataFrame--")
print(df)
```

Accessing DatFrame data:

We know that in our DataFrame data is saved in 2D form of Rows & Columns
if we want only one row/one column we use:
1. One column: df.['column_name']
2. one Row: .iloc[]
ex: df.iloc[0] indicates row in zero index.

```python
import pandas as pd
chemical_data = {
    "Nanoparticle": ["ZrO2", "TiO2", "SiO2"],
    "Size_nm": [25,45,60],
    "stability": ["High","Medium", "High"]
}
df = pd.DataFrame(chemical_data)
sizes_only = df['Size_nm']
first_row = df.iloc[0]
print("--Pandas lab: Accessing data-- ")
print("Total sizes of column:\n", sizes_only)
print("\n first chemical information (Row 0):\n", first_row)
```

Finding null values/missing values of data by using  a function in pandas
null values= .isnull().sum()  

```python
import pandas as pd
chemical_data = {
    "Nanoparticle": ["ZrO2", "TiO2", "SiO2", "Al2O3"],
    "Size_nm": [25, None, 45,60],
    "stability": ["High","Medium", "High", "Low"]
}
df = pd.DataFrame(chemical_data)
missing_values_count = df.isnull().sum()
print("--Pandas lab: Finding missing data--")
print("our chemical table:\n", df)
print("\n null values present in each column:\n", missing_values_count)
```

Dta cleaning:

In DataFrame null value is filled by our selcted number by using a function in pandas: .fillna()
generally to fill null in apertucular row\column we take average of entire row\column of thet perticula null and fill that average\mean value in that is the best optin used by lage undustries.

```python
import pandas as pd
chemical_data = {
    "Nanoparticles": ["ZrO2","TiO2", "SiO2", "Al2O3"],
    "Size_nm": [25.0, None, 45.0, 60.0],
    "stability": ["High", "Medium", "High", "Low"]
}
df = pd.DataFrame(chemical_data)
mean_size = df['Size_nm'].mean()
df['Size_nm'] = df['Size_nm'].fillna(mean_size)
print("--Pandas lab: Data cleaning with mean--")
print("calculated mean size (mean size):", mean_size)
print("\n new chemical table after cleaning:\n", df)
```

Pandas DataFiltering:

If in our labdata thousands of chemicals are present in that we want to take only requred data by using condition is called data filtering.
codition used: Ex: df[df[Size_nm] > 30] means here we collect data of above 30nm size particles from large data.

```python
import pandas as pd
chemical_data = {
    "Nanoparticles": ["ZrO2", "TiO2", "SiO2", "Al2O3"],
    "Size_nm": [25.0, 43.33, 45.0, 60.0],
    "Stability": ["High", "Medium", "High", "Low"]
}
df = pd.DataFrame(chemical_data)
large_particles = df[df['Size_nm'] > 40]
print("Pands lab: Datafiltering")
print("Chemicals size with > 40nm:\n", large_particles)
```
